# in-context learning, mini

short notebook on the simplest possible icl experiment: classify reviews positive/negative with k examples in the prompt and see how accuracy scales with k.

```python
# fake the scoring by templating the prompt
examples = [
    ('the soup was lukewarm but i ate it', 'neutral'),
    ('best birthday cake i ever had', 'positive'),
    ('rude host, never going back', 'negative'),
    ('food was fine, prices fair', 'neutral'),
]

def build_prompt(test_text, k=2):
    parts = []
    for t, lbl in examples[:k]:
        parts.append(f'Review: {t}\nSentiment: {lbl}\n')
    parts.append(f'Review: {test_text}\nSentiment:')
    return '\n'.join(parts)

print(build_prompt('the staff was charming and the wine list deep', k=3))
```

with k=0 my accuracy on a 30 example holdout was ~70 percent on davinci. k=2 -> ~85, k=4 -> ~88, k=8 plateaued. consistent with what most icl papers say.
