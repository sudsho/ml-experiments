# pydantic v2 migration

v2 came out june 2023. mostly a rust-backed rewrite, ~5-10x faster validation. some api changes that bite though.

```python
from pydantic import BaseModel, field_validator

class User(BaseModel):
    name: str
    age: int

    @field_validator('age')
    @classmethod
    def must_be_positive(cls, v):
        if v < 0: raise ValueError('age must be positive')
        return v
```
