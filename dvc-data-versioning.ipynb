# dvc data versioning

problem: training data changes, models change, but git can't track 5GB csvs. dvc is git for data.

## setup (cli, not in notebook)
```
pip install 'dvc[s3]'
dvc init
dvc remote add -d storage s3://my-bucket/dvc
dvc add data/raw/big.csv
git add data/raw/big.csv.dvc .gitignore
dvc push
```

````python
# dvc pipeline stages can be defined in dvc.yaml
import yaml
stages = {
    'stages': {
        'preprocess': {
            'cmd': 'python src/preprocess.py',
            'deps': ['src/preprocess.py', 'data/raw/big.csv'],
            'outs': ['data/processed/clean.parquet'],
        },
        'train': {
            'cmd': 'python src/train.py',
            'deps': ['src/train.py', 'data/processed/clean.parquet'],
            'outs': ['models/model.pkl'],
            'metrics': ['metrics.json'],
        },
    }
}
print(yaml.dump(stages))
````
