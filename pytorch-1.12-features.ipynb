## fx symbolic trace on a tiny module

not trying to do real graph rewrite yet, just see what it spits out.

```python
import torch.nn as nn
from torch.fx import symbolic_trace

class M(nn.Module):
    def __init__(self):
        super().__init__()
        self.l = nn.Linear(8, 4)
    def forward(self, x):
        return self.l(x).relu()

m = M()
print(symbolic_trace(m).graph)
```

verdict so far: the linalg surface is the win for me, channels_last is nice but only matters on real gpus.
