# gradio intro

streamlit was great in 2020. gradio is much faster for ML demos because the interface auto-generates from the function signature. trying it.

```python
import gradio as gr

def classify(text):
    return {'positive': 0.7, 'negative': 0.3}

demo = gr.Interface(fn=classify, inputs='text', outputs='label')
# demo.launch()  # runs a local server
```

## image input

```python
import numpy as np
def invert(img):
    return 255 - img

demo2 = gr.Interface(fn=invert, inputs='image', outputs='image')
# demo2.launch()
```

## streamlit vs gradio in my notes
- streamlit: full app feel, sidebars, multipage, persistent state, more code
- gradio: function -> ui, much less code, perfect for sharing one model

for portfolio demos i'm switching to gradio.
