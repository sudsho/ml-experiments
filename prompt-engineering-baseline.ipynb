## zero-shot

```
Classify the sentiment of this review (positive / negative / neutral):

Review: I waited 40 minutes and the coffee was cold.
Sentiment:
```

works ok for sentiment, fails on anything subtle.

## few-shot

```
Review: best burger in town.
Sentiment: positive

Review: the fries were soggy but the staff was nice.
Sentiment: neutral

Review: never coming back, host was rude.
Sentiment: negative

Review: the soup was warm and the bread fresh.
Sentiment:
```

big jump in agreement vs my own labels. 3-4 examples seems like a sweet spot.

## chain-of-thought

append `let's think step by step.` for arithmetic / logic. lifts a basic word-problem benchmark from ~30 percent to ~60 percent in my tiny test set. paraphrasing, not real eval.

## role priming

```
You are a senior python developer. ...
```

useful when i want code in a particular style, less so for facts.
