# tensorboard for tracking

basic torch + tensorboard. lightning gives this for free but i want to know what's underneath.

```python
from torch.utils.tensorboard import SummaryWriter
import torch
import torch.nn as nn

writer = SummaryWriter('runs/exp1')
for step in range(100):
    writer.add_scalar('loss/train', 1.0/(step+1), step)
writer.close()
```

## logging images and histograms

```python
import torchvision
img = torch.randn(3, 32, 32)
writer = SummaryWriter('runs/exp2')
writer.add_image('sample', img, 0)
for name, p in nn.Linear(10, 10).named_parameters():
    writer.add_histogram(name, p, 0)
writer.close()
```

## quick notes
- run `tensorboard --logdir runs` to view
- scalar/image/histogram/text all work
- comparing runs side by side is the killer feature
