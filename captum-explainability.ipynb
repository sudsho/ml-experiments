# captum: pytorch explainability

shap and lime were great in 2020. for torch models specifically, captum is more native. trying integrated gradients on a small classifier.

```python
import torch
import torch.nn as nn
from captum.attr import IntegratedGradients

class Tiny(nn.Module):
    def __init__(self):
        super().__init__()
        self.l1 = nn.Linear(20, 32)
        self.l2 = nn.Linear(32, 2)
    def forward(self, x):
        return self.l2(torch.relu(self.l1(x)))

m = Tiny().eval()
x = torch.randn(4, 20, requires_grad=True)
ig = IntegratedGradients(m)
attr = ig.attribute(x, target=1)
attr.shape
```
