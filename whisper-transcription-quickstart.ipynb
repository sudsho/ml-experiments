- tiny: very fast on cpu, occasional drops on technical terms
- base: surprisingly good for everyday english
- small: my sweet spot on a laptop cpu. ~1.6x realtime
- medium: noticeably better on accented speech, ~3.5x realtime on cpu so not realtime
