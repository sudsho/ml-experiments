# pytorch intro

first pytorch notebook. just tensors and autograd.


```python
import torch
print(torch.__version__)
print('cuda:', torch.cuda.is_available())
```

```python
# basic tensor ops
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])
print(x + y)
print(x * y)
print(x.dot(y))
```

```python
# autograd
x = torch.tensor(2.0, requires_grad=True)
y = x**3 + 2*x
y.backward()
print('dy/dx at x=2:', x.grad.item())  # 3*4 + 2 = 14
```

```python
# in-place vs out-of-place
x = torch.tensor([1.,2.,3.])
y = x.add(1)
print(x, y)
x.add_(1)
print(x)
```

```python
# tensor reshapes
z = torch.arange(12)
print(z.view(3,4))
print(z.reshape(2,6))
print(z.unsqueeze(0).shape)
```

```python
# numpy interop
import numpy as np
a = np.arange(6).reshape(2,3)
t = torch.from_numpy(a)
print(t)
print(t.numpy())
```

```python
# simple mlp
import torch.nn as nn
mlp = nn.Sequential(nn.Linear(10, 32), nn.ReLU(), nn.Linear(32, 2))
x = torch.randn(8, 10)
print(mlp(x).shape)
```

```python
# manual gradient descent on a quadratic
x = torch.tensor(5.0, requires_grad=True)
lr = 0.1
for _ in range(20):
    loss = (x - 2)**2
    loss.backward()
    with torch.no_grad():
        x -= lr * x.grad
        x.grad.zero_()
print('final x:', x.item())
```

```python
# device
dev = 'cuda' if torch.cuda.is_available() else 'cpu'
print(dev)
x = torch.randn(3,3).to(dev)
print(x.device)
```

### compared to keras: pytorch is more verbose for vanilla cases but the explicit loop helps when debugging.


### note: torch.no_grad() is essential during eval to skip grad tracking.
