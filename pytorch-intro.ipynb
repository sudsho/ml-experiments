# pytorch intro

first pytorch notebook. just tensors and autograd.


```python
import torch
print(torch.__version__)
print('cuda:', torch.cuda.is_available())
```

```python
# basic tensor ops
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])
print(x + y)
print(x * y)
print(x.dot(y))
```

```python
# autograd
x = torch.tensor(2.0, requires_grad=True)
y = x**3 + 2*x
y.backward()
print('dy/dx at x=2:', x.grad.item())  # 3*4 + 2 = 14
```

```python
# in-place vs out-of-place
x = torch.tensor([1.,2.,3.])
y = x.add(1)
print(x, y)
x.add_(1)
print(x)
```

```python
# tensor reshapes
z = torch.arange(12)
print(z.view(3,4))
print(z.reshape(2,6))
print(z.unsqueeze(0).shape)
```

```python
# numpy interop
import numpy as np
a = np.arange(6).reshape(2,3)
t = torch.from_numpy(a)
print(t)
print(t.numpy())
```
