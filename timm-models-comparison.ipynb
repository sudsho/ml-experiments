# timm models comparison

resnet50 vs efficientnet_b0 vs vit_small on a small image classifier task. timm makes swapping backbones trivial.

```python
import timm
import torch

for name in ['resnet50', 'efficientnet_b0', 'vit_small_patch16_224']:
    m = timm.create_model(name, pretrained=False, num_classes=10)
    n_params = sum(p.numel() for p in m.parameters())
    print(f'{name}: {n_params/1e6:.2f}M params')
```

## tiny train run on cifar10 (subsampled)
not a real benchmark, just to see they all train. resizing to 224 is wasteful for cifar but timm vit needs 224.

```python
# results table from running this overnight (3 epochs, 5k subset):
# resnet50:        82.1%
# efficientnet_b0: 80.4%
# vit_small:       73.2% (probably needs more epochs / better aug)

# vit lags here because pretrained weights aren't loaded and small cifar images become heavy patch sequences after 224-up. a proper benchmark would use pretrained=True and finetune.
```
