# timm models comparison

resnet50 vs efficientnet_b0 vs vit_small on a small image classifier task. timm makes swapping backbones trivial.

```python
import timm
import torch

for name in ['resnet50', 'efficientnet_b0', 'vit_small_patch16_224']:
    m = timm.create_model(name, pretrained=False, num_classes=10)
    n_params = sum(p.numel() for p in m.parameters())
    print(f'{name}: {n_params/1e6:.2f}M params')
```
