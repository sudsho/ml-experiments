# PCA + dim reduction

PCA on iris and breast cancer. visualize 2D projection.

```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
%matplotlib inline

iris = load_iris()
Xs = StandardScaler().fit_transform(iris.data)
pca = PCA(n_components=2)
Z = pca.fit_transform(Xs)
plt.scatter(Z[:,0], Z[:,1], c=iris.target)
plt.xlabel('pc1'); plt.ylabel('pc2')
print('explained ratio:', pca.explained_variance_ratio_)
```

## variance vs n_components on cancer

```python
from sklearn.datasets import load_breast_cancer
import numpy as np
Xc = StandardScaler().fit_transform(load_breast_cancer().data)
p = PCA().fit(Xc)
plt.plot(np.cumsum(p.explained_variance_ratio_))
plt.xlabel('n components'); plt.ylabel('cumulative var')
```

## downstream: classify on top-2 PCs

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
Xs2 = PCA(n_components=2).fit_transform(StandardScaler().fit_transform(load_iris().data))
cross_val_score(LogisticRegression(solver='lbfgs', multi_class='auto', max_iter=300),
                Xs2, load_iris().target, cv=5).mean()
```
