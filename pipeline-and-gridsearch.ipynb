# pipeline + grid search

put a scaler and an estimator in a Pipeline, search over hyperparams.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline

X, y = load_breast_cancer(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, stratify=y, random_state=1)
```

```python
pipe = Pipeline([('scale', StandardScaler()), ('svc', SVC())])
params = {
    'svc__C': [0.1, 1, 10],
    'svc__gamma': ['scale', 'auto'],
    'svc__kernel': ['rbf', 'linear']
}
gs = GridSearchCV(pipe, params, cv=5, n_jobs=-1)
gs.fit(Xtr, ytr)
print(gs.best_params_)
print('test', gs.score(Xte, yte))
```

## random forest variant

```python
from sklearn.ensemble import RandomForestClassifier
rf_pipe = Pipeline([('rf', RandomForestClassifier(random_state=0))])
rf_params = {
    'rf__n_estimators': [50, 100, 200],
    'rf__max_depth': [None, 5, 10],
}
gs2 = GridSearchCV(rf_pipe, rf_params, cv=5, n_jobs=-1)
gs2.fit(Xtr, ytr)
print(gs2.best_params_)
print('test', gs2.score(Xte, yte))
```

## RandomizedSearchCV variant

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import loguniform
rs = RandomizedSearchCV(pipe,
    {'svc__C': loguniform(0.01, 100), 'svc__gamma': loguniform(0.001, 1)},
    n_iter=20, cv=5, n_jobs=-1, random_state=0)
rs.fit(Xtr, ytr)
rs.best_params_, rs.best_score_
```

## takeaway
Grid is brute-force, RandomizedSearch is faster for big spaces.
