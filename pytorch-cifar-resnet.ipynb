# pytorch cifar resnet

transfer learning preview. resnet18 on cifar10, fine-tune the head.


```python
import torch
import torch.nn as nn
import torchvision
import torchvision.transforms as T
from torch.utils.data import DataLoader

tfm = T.Compose([
    T.Resize(64),
    T.ToTensor(),
    T.Normalize((0.4914, 0.4822, 0.4465), (0.247, 0.243, 0.261)),
])
tr = torchvision.datasets.CIFAR10('data/cifar', train=True, download=True, transform=tfm)
te = torchvision.datasets.CIFAR10('data/cifar', train=False, download=True, transform=tfm)
print(len(tr), len(te))
```

```python
net = torchvision.models.resnet18(pretrained=True)
net.fc = nn.Linear(net.fc.in_features, 10)
print(sum(p.numel() for p in net.parameters() if p.requires_grad))
```

### pretrained imagenet weights, just retrain the last fc layer first.
