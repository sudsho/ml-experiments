# seaborn EDA

seaborn is way prettier than matplotlib for quick eda. trying it on iris.

```python
import seaborn as sns
import pandas as pd
from sklearn.datasets import load_iris
%matplotlib inline

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
df.head()
```

```python
sns.pairplot(df, hue='species')
```

## boxplots by species

```python
sns.boxplot(x='species', y='petal length (cm)', data=df)
```

## correlation heatmap

```python
corr = df.iloc[:, :-1].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
```

## violinplot

```python
sns.violinplot(x='species', y='sepal length (cm)', data=df)
```

## quick try with wine quality (sklearn doesn't have it; use a tiny synthetic stand-in)

```python
from sklearn.datasets import load_wine
wd = load_wine()
wdf = pd.DataFrame(wd.data, columns=wd.feature_names)
wdf['target'] = pd.Categorical.from_codes(wd.target, wd.target_names)
sns.scatterplot(x='alcohol', y='color_intensity', hue='target', data=wdf)
```
