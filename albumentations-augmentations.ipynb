# albumentations

faster augmentations than torchvision (allegedly). going to time them.

```python
import albumentations as A
import cv2
import numpy as np

tfm = A.Compose([
    A.RandomResizedCrop(224, 224),
    A.HorizontalFlip(p=0.5),
    A.ColorJitter(0.2, 0.2, 0.2, 0.05),
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
])
x = np.random.randint(0, 255, (256, 256, 3), dtype=np.uint8)
out = tfm(image=x)['image']
out.shape
```

## benchmark vs torchvision

```python
import time
from torchvision import transforms as T
tv = T.Compose([T.ToPILImage(), T.RandomResizedCrop(224), T.RandomHorizontalFlip(), T.ColorJitter(0.2,0.2,0.2,0.05), T.ToTensor(), T.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])])

N = 200
t0 = time.time()
for _ in range(N):
    tfm(image=x)
print('alb:', time.time()-t0)
t0 = time.time()
for _ in range(N):
    tv(x)
print('tv:', time.time()-t0)
```
