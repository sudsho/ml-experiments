# glove embeddings

loading pretrained glove vectors and doing classic nn analogies.


```python
import numpy as np

def load_glove(path, dim=100, limit=None):
    word2vec = {}
    with open(path, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f):
            if limit and i >= limit:
                break
            parts = line.rstrip().split(' ')
            word2vec[parts[0]] = np.array(parts[1:], dtype=np.float32)
    return word2vec

# expects data/glove.6B.100d.txt
# wv = load_glove('data/glove.6B.100d.txt', dim=100, limit=50000)
```

```python
wv = load_glove('data/glove.6B.100d.txt', dim=100, limit=50000)
print(len(wv))
```

```python
def cosine(a, b):
    return (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))

print(cosine(wv['king'], wv['queen']))
print(cosine(wv['cat'], wv['car']))
```

```python
# nearest neighbors
words = list(wv.keys())
mat = np.stack([wv[w] for w in words])
mat = mat / np.linalg.norm(mat, axis=1, keepdims=True)

def nn_of(word, k=10):
    v = wv[word] / np.linalg.norm(wv[word])
    sims = mat @ v
    idx = sims.argsort()[::-1][:k]
    return [(words[i], sims[i]) for i in idx]

for w in ['paris','python','sad']:
    print(w, '->', nn_of(w, 5))
```
