# ray tune

used optuna in 2020. ray tune lets you do the same kind of search but distributed. trying on a quick mnist-like sweep.

```python
import ray
from ray import tune

def objective(config):
    # pretend training
    score = (config['lr'] - 0.001) ** 2 + (config['bs'] - 64) ** 2 / 1000
    tune.report(score=score)

ray.init(ignore_reinit_error=True)
result = tune.run(objective, config={'lr': tune.loguniform(1e-5, 1e-1), 'bs': tune.choice([16, 32, 64, 128])}, num_samples=20)
```

## ASHA scheduler
async hyperband. early-stops bad trials. nice for nontrivial training time.

```python
from ray.tune.schedulers import ASHAScheduler
asha = ASHAScheduler(metric='score', mode='min', max_t=100, grace_period=10)
# run with scheduler=asha
```

## verdict
optuna is simpler for single-machine. ray tune for cluster + early stopping. learn both.
