## boolean masks

```python
# argsort vs sort
v = np.array([3, 1, 4, 1, 5])
print(np.sort(v))
print(np.argsort(v))
```

## linalg

```python
A = np.array([[1,2],[3,4]])
print(np.linalg.inv(A))
print(np.linalg.eigvals(A))
```
