# jax vs pytorch microbench

tiny matmul + grad timings on cpu, just to feel the gap. not a real benchmark.

```python
import time
import torch

def bench_torch(n=512, iters=20):
    a = torch.randn(n, n); b = torch.randn(n, n)
    t0 = time.time()
    for _ in range(iters):
        c = a @ b
    return (time.time() - t0) / iters

print(bench_torch())
```
