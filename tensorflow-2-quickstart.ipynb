# tensorflow 2 quickstart

following the official tf2 tutorial. tf.keras feels much cleaner than tf1.


```python
import tensorflow as tf
print(tf.__version__)
# eager by default in tf2
print(tf.executing_eagerly())
```

```python
(x_tr, y_tr), (x_te, y_te) = tf.keras.datasets.mnist.load_data()
x_tr = x_tr / 255.0
x_te = x_te / 255.0
print(x_tr.shape, y_tr.shape)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10),
])
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])
```

```python
model.fit(x_tr, y_tr, epochs=5, batch_size=128, validation_split=0.1)
```

```python
test_loss, test_acc = model.evaluate(x_te, y_te)
print('test acc:', test_acc)
```

### test acc ~0.978. tf.keras feels close to keras-only.


```python
# functional api
inp = tf.keras.Input(shape=(28,28))
x = tf.keras.layers.Flatten()(inp)
x = tf.keras.layers.Dense(128, activation='relu')(x)
x = tf.keras.layers.Dropout(0.2)(x)
out = tf.keras.layers.Dense(10)(x)
model2 = tf.keras.Model(inp, out)
model2.summary()
```

```python
# tf.data
ds = tf.data.Dataset.from_tensor_slices((x_tr, y_tr)).shuffle(1000).batch(128)
for batch in ds.take(1):
    print(batch[0].shape, batch[1].shape)
```

```python
# subclassing
class MyModel(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.flat = tf.keras.layers.Flatten()
        self.d1 = tf.keras.layers.Dense(128, activation='relu')
        self.d2 = tf.keras.layers.Dense(10)
    def call(self, x):
        return self.d2(self.d1(self.flat(x)))
m3 = MyModel()
```
