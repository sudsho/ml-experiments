# great expectations: data quality

the pain: silently bad data poisons training. ge lets you write expectations against a dataframe and fail loudly.

```python
import great_expectations as ge
import pandas as pd

df = pd.DataFrame({
    'age': [25, 33, -1, 47, 200, 18],
    'income': [50000, 70000, None, 90000, 120000, 30000],
    'state': ['CA', 'TX', 'NY', 'KY', 'XX', 'CA'],
})
gdf = ge.from_pandas(df)
```

## a few expectations

```python
print(gdf.expect_column_values_to_be_between('age', min_value=0, max_value=120))
print(gdf.expect_column_values_to_not_be_null('income'))
print(gdf.expect_column_values_to_be_in_set('state', ['CA','TX','NY','KY','FL']))
```
