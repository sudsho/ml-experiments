# cross validation strategies

KFold vs StratifiedKFold vs LeaveOneOut. on iris.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import KFold, StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression

iris = load_iris()
X, y = iris.data, iris.target
lr = LogisticRegression(solver='lbfgs', multi_class='auto', max_iter=200)
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=0)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
print('kfold', cross_val_score(lr, X, y, cv=kf).mean())
print('stratified', cross_val_score(lr, X, y, cv=skf).mean())
```

## LeaveOneOut

```python
from sklearn.model_selection import LeaveOneOut
loo = LeaveOneOut()
# slow on full iris but fine for demo
scores = cross_val_score(lr, X[:50], y[:50], cv=loo)
print('mean', scores.mean(), 'n splits', len(scores))
```

## shuffle matters when classes are sorted

```python
from sklearn.model_selection import KFold
no_shuf = KFold(n_splits=5, shuffle=False)
yes_shuf = KFold(n_splits=5, shuffle=True, random_state=0)
print('no shuffle ', cross_val_score(lr, X, y, cv=no_shuf).mean())
print('shuffled  ', cross_val_score(lr, X, y, cv=yes_shuf).mean())
```

## nested CV

```python
from sklearn.model_selection import GridSearchCV
inner = StratifiedKFold(n_splits=3, shuffle=True, random_state=0)
outer = StratifiedKFold(n_splits=3, shuffle=True, random_state=1)
g = GridSearchCV(lr, {'C':[0.1,1,10]}, cv=inner)
scores = cross_val_score(g, X, y, cv=outer)
print('outer scores', scores)
print('mean', scores.mean())
```
