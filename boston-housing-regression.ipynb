# boston housing regression

old sklearn dataset. compare LinearReg vs Ridge vs RF.

```python
from sklearn.datasets import load_boston
import pandas as pd

boston = load_boston()
X = pd.DataFrame(boston.data, columns=boston.feature_names)
y = boston.target
X.head()
```

```python
from sklearn.model_selection import train_test_split
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
lr = LinearRegression()
lr.fit(Xtr, ytr)
pred = lr.predict(Xte)
print('rmse', mean_squared_error(yte, pred) ** 0.5)
print('r2', r2_score(yte, pred))
```

```python
from sklearn.linear_model import Ridge
ridge = Ridge(alpha=1.0)
ridge.fit(Xtr, ytr)
p2 = ridge.predict(Xte)
print('rmse', mean_squared_error(yte, p2) ** 0.5)
print('r2', r2_score(yte, p2))
```

```python
from sklearn.ensemble import RandomForestRegressor
rfreg = RandomForestRegressor(n_estimators=200, random_state=0)
rfreg.fit(Xtr, ytr)
p3 = rfreg.predict(Xte)
print('rmse', mean_squared_error(yte, p3) ** 0.5)
print('r2', r2_score(yte, p3))
```

## note
boston had a $50k price ceiling — explains the cluster of points at y=50.

## try log target

```python
import numpy as np
ytr_l = np.log1p(ytr); yte_l = np.log1p(yte)
lr.fit(Xtr, ytr_l)
p = np.expm1(lr.predict(Xte))
print('rmse on original scale:', mean_squared_error(yte, p) ** 0.5)
```
