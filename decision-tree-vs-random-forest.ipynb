# decision tree vs random forest

show how a single deep tree overfits, RF doesn't as much.

```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
X, y = make_moons(n_samples=500, noise=0.3, random_state=0)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
dt = DecisionTreeClassifier(random_state=0)
rf = RandomForestClassifier(n_estimators=200, random_state=0)
dt.fit(Xtr, ytr); rf.fit(Xtr, ytr)
print('dt:', dt.score(Xtr, ytr), dt.score(Xte, yte))
print('rf:', rf.score(Xtr, ytr), rf.score(Xte, yte))
```

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
def plot_db(model, X, y, title):
    h = .02
    x_min, x_max = X[:,0].min()-.5, X[:,0].max()+.5
    y_min, y_max = X[:,1].min()-.5, X[:,1].max()+.5
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=.4)
    plt.scatter(X[:,0], X[:,1], c=y, edgecolors='k')
    plt.title(title)
fig = plt.figure(figsize=(10,4))
plt.subplot(1,2,1); plot_db(dt, Xtr, ytr, 'tree')
plt.subplot(1,2,2); plot_db(rf, Xtr, ytr, 'forest')
plt.tight_layout()
```
