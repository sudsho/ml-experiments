wandb is fine but mlflow is the standard at most jobs i'm seeing. learning the api.

```python
import mlflow
import mlflow.sklearn
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

X, y = load_iris(return_X_y=True)
Xt, Xv, yt, yv = train_test_split(X, y, test_size=0.2, random_state=0)

mlflow.set_experiment('iris-logreg')
```

## logging a run

```python
for C in [0.1, 1.0, 10.0]:
    with mlflow.start_run(run_name=f'C={C}'):
        m = LogisticRegression(C=C, max_iter=200)
        m.fit(Xt, yt)
        acc = m.score(Xv, yv)
        mlflow.log_param('C', C)
        mlflow.log_metric('val_acc', acc)
        mlflow.sklearn.log_model(m, 'model')
        print(f'C={C} acc={acc:.3f}')
```
