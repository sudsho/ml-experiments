# mnist digits with sklearn

sklearn ships a small (~1700) digits dataset, not the full keras MNIST. good enough for warmup.

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
%matplotlib inline

digits = load_digits()
print(digits.data.shape)
print(digits.target.shape)
```

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='gray_r')
    ax.set_title(digits.target[i])
    ax.axis('off')
plt.tight_layout()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
Xtr, Xte, ytr, yte = train_test_split(digits.data, digits.target, test_size=0.25, random_state=0)
lr = LogisticRegression(solver='lbfgs', multi_class='auto', max_iter=2000)
lr.fit(Xtr, ytr)
lr.score(Xte, yte)
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
pred = lr.predict(Xte)
cm = confusion_matrix(yte, pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
```
