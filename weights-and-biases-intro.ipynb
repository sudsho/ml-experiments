team at work uses wandb. trying it on a toy run.

```python
import wandb
import torch
import torch.nn as nn
import torch.nn.functional as F

wandb.init(project='ml-experiments-2021', name='wandb-intro', config={'lr': 1e-3, 'bs': 64})
m = nn.Linear(10, 2)
opt = torch.optim.Adam(m.parameters(), lr=wandb.config.lr)
```

## tracking a fake training loop

```python
for step in range(50):
    x = torch.randn(64, 10)
    y = torch.randint(0, 2, (64,))
    logits = m(x)
    loss = F.cross_entropy(logits, y)
    opt.zero_grad()
    loss.backward()
    opt.step()
    wandb.log({'loss': loss.item(), 'step': step})
wandb.finish()
```
