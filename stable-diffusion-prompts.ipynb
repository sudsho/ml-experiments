## quick takeaways from a couple hundred runs

- guidance 5-8 looks best for most things. above 12 you get artifacts.
- 30 steps is fine for most prompts on the v1.4 weights.
- adding a medium (oil painting, watercolor, photo) early in the prompt helps a lot.
- negative prompts are crucial for hands and faces.
