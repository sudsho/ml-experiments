# streamlit 1.10+ multipage app

streamlit got native multipage in 1.10. before that we glued things together with a session_state hack. this is the new clean way.

## file layout
```
app/
  Home.py
  pages/
    1_EDA.py
    2_Train.py
    3_Predict.py
```

````python
home_py = '''
import streamlit as st
st.set_page_config(page_title="ml playground", layout="wide")
st.title("ml playground")
st.write("pick a page on the left.")
'''
print(home_py)
````

````python
predict_py = '''
import streamlit as st
import joblib, numpy as np

st.title("predict")
file = st.file_uploader("upload model.joblib", type=["joblib"])
features = st.text_input("features (csv)", "5.1,3.5,1.4,0.2")
if file and features:
    m = joblib.load(file)
    arr = np.array([float(x) for x in features.split(",")]).reshape(1, -1)
    st.write("prediction", int(m.predict(arr)[0]))
'''
print(predict_py)
````
