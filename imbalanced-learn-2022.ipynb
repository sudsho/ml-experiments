# imbalanced-learn, 2022 refresh

revisiting the smote/rose/tomek toolkit on a heavy class imbalance problem. comparing baseline vs resampled vs class_weight + threshold tuning.

```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, average_precision_score
from sklearn.model_selection import train_test_split
import numpy as np

X, y = make_classification(n_samples=20000, weights=[0.97, 0.03], random_state=0)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, stratify=y, random_state=0)
print('class balance:', y.mean())
```

```python
# baseline
base = LogisticRegression(max_iter=500).fit(X_tr, y_tr)
p = base.predict_proba(X_te)[:, 1]
print('auc', roc_auc_score(y_te, p), 'ap', average_precision_score(y_te, p))
```

rule of thumb i landed on: try `class_weight='balanced'` first, then threshold tuning, only reach for resampling when neither moves the needle. ap is more honest than auc on heavy imbalance.
