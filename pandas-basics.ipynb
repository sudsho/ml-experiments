```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'a': [1,2,3,4], 'b': [5,6,7,8]})
df.head()
```

```python
df.describe()
```

### filtering rows

```python
df[df.a > 2]
```

## groupby

```python
df2 = pd.DataFrame({
    'group': ['x','x','y','y','z'],
    'val': [1, 2, 3, 4, 5]
})
df2.groupby('group').mean()
```

## merge / join

```python
left = pd.DataFrame({'k':['a','b','c'], 'lv':[1,2,3]})
right = pd.DataFrame({'k':['b','c','d'], 'rv':[20,30,40]})
left.merge(right, on='k', how='left')
```

## pivot

```python
long = pd.DataFrame({
    'date':['2019-01','2019-01','2019-02','2019-02'],
    'metric':['a','b','a','b'],
    'value':[1,2,3,4],
})
long.pivot(index='date', columns='metric', values='value')
```

## apply

```python
df.apply(lambda r: r.a + r.b, axis=1)
```

## datetime

```python
ds = pd.date_range('2019-01-01', periods=10)
pd.DataFrame({'date': ds, 'val': range(10)}).set_index('date').resample('3D').sum()
```
