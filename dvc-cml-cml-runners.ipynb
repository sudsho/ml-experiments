# dvc + cml runners (cml 0.2+) for ci/cd in ml

wiring dvc data + cml runners into a github repo. cml comments on the pr with the metrics diff between branches. the part that took me longest was getting the runner labels right.

```python
# .github/workflows/cml.yml (kept here as a string so the file does not actually exist; it lives in the example repo)
workflow = '''
name: cml
on: [pull_request]
jobs:
  train-and-report:
    runs-on: [self-hosted, cml-runner]
    steps:
      - uses: actions/checkout@v3
      - uses: iterative/setup-cml@v1
      - name: train
        env:
          REPO_TOKEN: ${{ secrets.GITHUB_TOKEN }}
        run: |
          pip install -r requirements.txt
          dvc pull
          dvc repro
          echo "## metrics" > report.md
          dvc metrics diff --show-md main >> report.md
          cml-send-comment report.md
'''
print(workflow[:500])
```

## the takeaway

dvc + cml is the cheapest way i have found to get pr-driven ml ci. the metrics diff comment is the magic. on a real team i would still prefer mlflow + a dashboard, but for a small project this is plenty.
