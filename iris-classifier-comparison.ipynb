# iris classifier comparison

goal: try LogReg, DT, RF, SVM on iris and see what wins. classic sklearn.

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target
X.head()
```

```python
from sklearn.model_selection import train_test_split
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=42)
Xtr.shape, Xte.shape
```

## logistic regression

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(solver='lbfgs', multi_class='auto', max_iter=200)
lr.fit(Xtr, ytr)
lr.score(Xte, yte)
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(lr, X, y, cv=5)
scores, scores.mean()
```

## decision tree

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(Xtr, ytr)
dt.score(Xte, yte)
```

## random forest

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(Xtr, ytr)
rf.score(Xte, yte)
```

## SVM

```python
from sklearn.svm import SVC
svm = SVC(kernel='rbf', gamma='auto')
svm.fit(Xtr, ytr)
svm.score(Xte, yte)
```

## results

```python
results = {
    'LR':  lr.score(Xte, yte),
    'DT':  dt.score(Xte, yte),
    'RF':  rf.score(Xte, yte),
    'SVM': svm.score(Xte, yte),
}
import pandas as pd
pd.Series(results).sort_values(ascending=False)
```

## confusion matrix on best (RF)

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
cm = confusion_matrix(yte, rf.predict(Xte))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=load_iris().target_names, yticklabels=load_iris().target_names)
```

## refactor: model dict

```python
models = {
    'lr':  LogisticRegression(solver='lbfgs', multi_class='auto', max_iter=300),
    'dt':  DecisionTreeClassifier(random_state=42),
    'rf':  RandomForestClassifier(n_estimators=200, random_state=42),
    'svm': SVC(gamma='auto'),
}
from sklearn.model_selection import cross_val_score
for n, m in models.items():
    print(f'{n:5s}  {cross_val_score(m, X, y, cv=5).mean():.4f}')
```
